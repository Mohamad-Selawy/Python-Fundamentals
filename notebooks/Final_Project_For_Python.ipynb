Welcome to your second final project! This project focuses on building a simulated 'Online Shopping Market' using the Python fundamentals you've mastered. This will allow you to apply concepts such as:

*   **Object-Oriented Programming (OOP):** Classes, Objects, Attributes, Methods.

## Project Structure: Classes

We will define four main classes for this project, each representing a key component of an online shopping market.

### Class 1: `Product`

This class will represent an individual product available for sale in the market.

#### Attributes:
*   `product_id` (str): A unique identifier for the product (e.g., 'P001').
*   `name` (str): The name of the product.
*   `price` (float): The price of the product.
*   `stock` (int): The quantity of the product currently available.

#### Methods:
*   `__init__(self, product_id, name, price, stock)`: Constructor to initialize a new `Product` object. Ensure price is positive and stock is non-negative.
*   `get_info(self)`: Returns a string with the product's ID, name, price, and current stock.
*   `update_stock(self, quantity)`: Updates the stock by adding `quantity` (can be negative for sales). Ensures stock doesn't go below zero.
*   `is_available(self, quantity)`: Checks if a given `quantity` of the product is available in stock.

### Class 2: `User`

This class will represent a customer in our online shopping market.

#### Attributes:
*   `user_id` (str): A unique identifier for the user.
*   `username` (str): The user's chosen username.
*   `email` (str): The user's email address.
*   `shopping_cart` (ShoppingCart): An instance of the `ShoppingCart` class associated with this user.
*   `order_history` (list): A list of dictionaries, where each dictionary represents a placed order.

#### Methods:
*   `__init__(self, user_id, username, email)`: Constructor to initialize a new `User` object. Initializes `shopping_cart` as a new `ShoppingCart` instance and `order_history` as an empty list.
*   `display_info(self)`: Prints the user's ID, username, and email.
*   `add_to_cart(self, product, quantity)`: Adds a product to the user's shopping cart.
*   `remove_from_cart(self, product_id, quantity)`: Removes a product from the user's shopping cart.
*   `view_cart(self)`: Displays the contents and total price of the user's shopping cart.
*   `Empty_Cart(self)`: Remove all items from Cart.

*   `add_to_order_history(self, order_details)`: Adds a completed order's details to the `order_history`.

### Class 3: `ShoppingCart`

This class manages the items a user wishes to purchase.

#### Attributes:
*   `user_id` (str): The ID of the user to whom this cart belongs.
*   `items` (dict): A dictionary where keys are `Product` objects and values are the `quantity` of that product in the cart.

#### Methods:
*   `__init__(self, user_id)`: Constructor to initialize a new `ShoppingCart` for a given user.
*   `add_item(self, product, quantity)`: Adds a specified `quantity` of a `Product` to the cart. Checks for product availability.
*   `remove_item(self, product_id, quantity)`: Removes a product from the cart (or reduces its quantity).
*   `get_total(self)`: Calculates and returns the total price of all items in the cart.
*   `display_cart(self)`: Prints all items in the cart with their quantities and prices, and the total.
*   `clear_cart(self)`: Empties the shopping cart.

### Class 4: `OrderManager`

This class will handle the process of placing orders, managing the market's products, and tracking all transactions.

#### Attributes:
*   `products` (dict): A dictionary where keys are product IDs (str) and values are `Product` objects.
*   `users` (dict): A dictionary where keys are user IDs (str) and values are `User` objects.
*   `orders` (list): A list of dictionaries, where each dictionary represents a placed order with details like `user_id`, `items`, `total`, `order_id`.

#### Methods:
*   `__init__(self)`: Constructor to initialize the manager with empty `products`, `users`, and `orders`.
*   `add_product(self, product)`: Adds a `Product` object to the manager. Ensures product ID is unique.
*   `add_user(self, user)`: Adds a `User` object to the manager. Ensures user ID is unique.
*   `place_order(self, user_id)`: Processes the `ShoppingCart` of a specified user to create an order. This method should:
    *   Check if the user exists.
    *   Check if the cart is not empty.
    *   Verify stock availability for all items in the cart.
    *   Update the stock of each product in the market.
    *   Generate a unique `order_id`.
    *   Clear the user's shopping cart.
    *   Add the order to the `orders` list and the user's `order_history`.
*   `get_product_details(self, product_id)`: Returns the `Product` object for a given ID.
*   `get_user_details(self, user_id)`: Returns the `User` object for a given ID.
*   `display_all_products(self)`: Prints details of all products in the market.
*   `display_all_users(self)`: Prints details of all registered users.
*   `display_all_orders(self)`: Prints details of all placed orders.
