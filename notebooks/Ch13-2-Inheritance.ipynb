class childClassName(parentClass1, baseClass2, ...):

## Single Inheritance
- supported by almost all OOP langauges
<img src="https://github.com/Mohamad-Selawy/Python-Fundamentals/blob/main/notebooks/resources/Inheritance.png?raw=1" width="100%">

````python
class A():
    def __init__(self):
        self.a = "A"

    def printMe(self):
        print("A's printMe called!")
        print('a = {}'.format(self.a))

    def sayHi(self):
        print('{} says HI!'.format(self.a))
````

````python
obja = A()
obja.printMe()
obja.sayHi()
````

````python
# single inheritance
class B(A):
    def __init__(self):
        # must explictly invoke base classes constructors
        # to inherit properties/attributes
        A.__init__(self) # try commenting this out
        self.b = 'B'

    def update(self):
        print("Attributes before modifaction: {} and {}".format(self.a, self.b))
        self.a = 'AAA' #can modify inherited attributes
        print("Attributes after modification: {} and {}".format(self.a, self.b))

    # overrides inherited printMe
    def printMe(self):
        print("B's printMe called")
        print('a = {}'.format(self.a))
````

````python
objb = B()
# shows that A's properties are inherited by B
objb.update()
````

````python
# object a's properties are independent from object b's properties
print("obja's property a = {}".format(obja.a))
print("objb's property a = {}".format(objb.a))
````

````python
# B inherits A's sayHi()
# what is the output of the following?
objb.sayHi()
````

## Overriding
- child class can redefine method that are inherited from parent class with the same name
- e.g., printMe() method in class B overrides A's printMe
- A's printme can still be called
    - syntax
    ```
    ClassName.method(object)
    ```

````python
objb.printMe()
````

````python
A.printMe(obja)
````

````python
A.printMe(objb)
````

````python
# C inherits from B which inherits from A
class C(B):
    def __init__(self):
        B.__init__(self)
        self.c = 'C'

    def printMe(self):
        print("C's printMe called:")
        print("Attributes are {}, {}, {}".format(self.c, self.b, self.a))
````

````python
c1 = C()
c1.printMe()
````

````python
# sayHi() inherited from A
c1.sayHi()
````

````python
c1.update()
````

## Exercise 1: `Vehicle` and `Car`/`Truck` Inheritance

This exercise focuses on demonstrating **inheritance** using a standard Base Class (`Vehicle`) and subclasses (`Car` and `Truck`).

Follow these steps to complete the exercise:

1.  **Define a Base Class `Vehicle`**:
    *   Create a class named `Vehicle`.
    *   Declare two methods within `Vehicle`: `start()` and `stop()`. These methods can print a default message or contain default behaviors.

2.  **Define a Subclass `Car`**:
    *   Create a class named `Car` that inherits from `Vehicle`.
    *   Override the `start()` method for `Car` to print a message like "Car engine started."
    *   Override the `stop()` method for `Car` to print a message like "Car engine stopped."
    *   Add a new concrete method `drive()` to the `Car` class that prints "Car is driving."

3.  **Define a Subclass `Truck`**:
    *   Create a class named `Truck` that also inherits from `Vehicle`.
    *   Override the `start()` method for `Truck` to print a message like "Truck engine started."
    *   Override the `stop()` method for `Truck` to print a message like "Truck engine stopped."
    *   Add a new concrete method `load_cargo(weight)` to the `Truck` class. This method should take a `weight` argument and print a message indicating that the truck is loading cargo of that `weight`.

4.  **Demonstrate Usage**:
    *   Instantiate an object of the `Car` class.
    *   Call its `start()`, `drive()`, and `stop()` methods.
    *   Instantiate an object of the `Truck` class.
    *   Call its `start()`, `load_cargo()` (with an example weight), and `stop()` methods.

````python
# 1. Define a Base Class Vehicle
class Vehicle:
    def start(self):
        print("Vehicle starting...")

    def stop(self):
        print("Vehicle stopping...")
````

````python
# 2. Define a Concrete Class Car
class Car(Vehicle):
    def start(self):
        print("Car engine started.")

    def stop(self):
        print("Car engine stopped.")

    def drive(self):
        print("Car is driving.")
````

````python
# 3. Define a Concrete Class Truck
class Truck(Vehicle):
    def start(self):
        print("Truck engine started.")

    def stop(self):
        print("Truck engine stopped.")

    def load_cargo(self, weight):
        print(f"Truck is loading cargo weighing {weight} kg.")
````

### Demonstration of Usage

````python
# Demonstrate Usage of Car
print("--- Car Demonstration ---")
my_car = Car()
my_car.start()
my_car.drive()
my_car.stop()

print("\n--- Truck Demonstration ---")
# Demonstrate Usage of Truck
my_truck = Truck()
my_truck.start()
my_truck.load_cargo(5000)
my_truck.stop()
````
