### Better class example
- with constructor (`__init__`), instance attributes, and methods


````python
class Point():
    """
    Point class to represent and manipulate x and y in 2D coordinates
    """

    # constructor to customize the initial state of an object
    # 'self' refers to the instance being created
    def __init__(self, x=0, y=0):
        """Create a new point with given x and y coords"""
        self.x = x   # object attribute
        self.y = y   # object attribute

    def p(self):
        print('x = ',self.x)
        print('y = ',self.y)
````

````python
u = Point(5, 8)
u.p()
````

````python
t = Point()
t.p()
````

### exercise: add a method dist_from_origin() to Point class
- computes and returns the distance from the origin
- test the methods
- provides `__str__` overloaded method to represent objects as string
    - helps in printing objects

````python
class Point:
    """
    Point class represents and manipulates x,y coords
    """
    count = 0

    def __init__(self, xx=0, yy=0):
        """Create a new point with given x and y coords"""
        self.x = xx
        self.y = yy
        Point.count += 1

    def dist_from_origin(self):
        import math
        dist = math.sqrt( self.x**2 + self.y**2 )
        return f"the distance is : {dist}"

    def __str__(self):
        return f"( x = {self.x}, y = {self.y} )"
````

### Member access specifiers

- Python doesn't support `private`, `public` specifiers provided by C++, Java, etc.
- all the members are public by default
- however, you can use leading single `_` and double `__` underscores "convention" to treat members as private

````python
class Point:
    """
    Point class represents and manipulates x, y coordinates
    """
    __count = 100
    def __init__(self, xx=0, yy=0):
        """Create a new point with given x and y coords"""
        self.x = xx
        self.y = yy

    def dist_from_origin(self):
        import math
        return math.sqrt(self.x**2 + self.y**2)

    def move(self, xx, yy):
        """Move point to new coordinates, with type validation."""
        if isinstance(xx, (int, float)) and isinstance(yy, (int, float)):
            self.x = xx
            self.y = yy
        else:
            print("Error: x and y coordinates must be numbers.")

    def __str__(self):
        return f"(x = {self.x}, y = {self.y})"
````

## Composition
- class can include another class as a member
- let's say we want to represent a rectangle in a 2-D coordinates (XY plane)
- corner represents the top left point on a XY plane

````python
class Rectangle:
    """ A class to manufacture rectangle objects """

    def __init__(self, posn, w, h):
        """ Initialize rectangle at posn, with width w, height h """
        self.corner = posn
        self.width = w
        self.height = h

    def __str__(self):
        return  "({0}, {1}, {2})".format(self.corner, self.width, self.height)
````

````python
box = Rectangle(Point(0, 0), 100, 200)
bomb = Rectangle(Point(100, 80), 5, 10)    # In my video game
print("box: ", box)
print("bomb: ", bomb)
````

## Exercise 1: Student Management System

**Task:** Create a `Student` class that allows managing student details and their grades for various courses.

**Requirements:**
1.  **`Student` Class Definition:**
    -   **Attributes:** `student_id` (string), `name` (string), `major` (string).
    -   Initialize another attribute called `grades` as an empty dictionary in the constructor (`__init__`), where keys will be `course_name` (string) and values will be `grade` (float or int).

2.  **Constructor (`__init__`)**: Define the constructor to accept `student_id`, `name`, and `major`.

3.  **String Representation (`__str__`)**: Implement the `__str__` method to return a user-friendly string displaying the student's ID, name, and major (e.g., "Student ID: [ID], Name: [Name], Major: [Major]").

4.  **Attribute Handling Method (`add_grade`)**: Create a method `add_grade(course_name, grade)`:
    -   This method should take a `course_name` (string) and a `grade` (float or int) as input.
    -   It should add or update the grade for the specified course in the `grades` dictionary attribute.
    -   Add a print statement to confirm the grade was added/updated.

5.  **Additional Method (`get_gpa`)**: Create a method `get_gpa()`:
    -   This method should calculate and return the average of all grades stored in the `grades` dictionary.
    -   If the `grades` dictionary is empty, it should return 0.0 or print a message indicating no grades are available.

**Demonstration Steps:**
1.  Create an instance of the `Student` class.
2.  Use the `add_grade` method multiple times to add grades for different courses.
3.  Print the `Student` object to see its `__str__` representation.
4.  Call the `get_gpa` method and print the calculated GPA.

````python
class Student:
    """
    A class to represent a student in a management system.
    """

    def __init__(self, student_id, name, major):
        self.student_id = student_id
        self.name = name
        self.major = major
        self.grades = {}

    def __str__(self):
        return f"Student ID: {self.student_id}, Name: {self.name}, Major: {self.major}"

    def add_grade(self, course_name, grade):
        self.grades[course_name] = grade
        print(f"Grade for {course_name} ({grade}) added/updated for {self.name}.")

    def get_gpa(self):
        if not self.grades:
            print(f"No grades available for {self.name}.")
            return 0.0
        total_grades = sum(self.grades.values())
        return total_grades / len(self.grades)

    def display_grades(self):
        if not self.grades:
            print(f"{self.name} has no grades recorded.")
            return
        print(f"\nGrades for {self.name}:")
        for course, grade in self.grades.items():
            print(f"  - {course}: {grade}")
````

### Demonstration of the `Student` class

````python
# 1. Create an instance of the Student class
student1 = Student(student_id='S1001', name='Alice Smith', major='Computer Science')
print(f"Created student: {student1.name}")

# 2. Use the add_grade method multiple times to add grades for different courses
student1.add_grade('Programming I', 92.5)
student1.add_grade('Calculus', 88)
student1.add_grade('Data Structures', 95)
student1.add_grade('Programming I', 93) # Update existing grade

# 3. Print the Student object to see its __str__ representation
print("\nStudent Information:")
print(student1)

# Use the new display_grades method
student1.display_grades()

# 4. Call the get_gpa method and print the calculated GPA
gpa = student1.get_gpa()
print(f"\n{student1.name}'s GPA: {gpa:.2f}")


# Demonstrate with a student with no grades
student2 = Student(student_id='S1002', name='Bob Johnson', major='Physics')
print(f"\nCreated student: {student2.name}")
print(student2)
student2.display_grades()
gpa2 = student2.get_gpa()
print(f"{student2.name}'s GPA: {gpa2:.2f}")
````
